```python
import numpy as np
import tensorflow as tf
from tensorflow import keras
import matplotlib.pyplot as plt
# Importiamo i dati
(x_train, y_train), (x_test, y_test) = keras.datasets.cifar10.load_data()

# Normalizzazione
x_train = x_train/255
x_test = x_test/255

Nlabels = np.max(y_train) + 1           # In questo caso le classi sono {0, ..., Nlabels - 1}

```

```python
histories = {}
for Nepochs, deep in [(6, False), (12, False), (6, True)]:
    model = keras.models.Sequential()

    # Convolutional layer
    model.add(keras.layers.Input(shape = (32, 32, 3)))
    model.add(keras.layers.Conv2D(64, (3,3)))
    model.add(keras.layers.Activation("relu"))

    # Pooling layer
    model.add(keras.layers.MaxPooling2D(pool_size = (2,2)))

    if deep:
        # Convolutional layer
        model.add(keras.layers.Conv2D(256, (3,3)))
        model.add(keras.layers.Activation("relu"))

        # Pooling layer
        model.add(keras.layers.MaxPooling2D(pool_size = (2,2)))

        # Convolutional layer
        model.add(keras.layers.Conv2D(90, (3,3)))
        model.add(keras.layers.Activation("relu"))

        # Pooling layer
        model.add(keras.layers.MaxPooling2D(pool_size = (2,2)))

    # Fully connected layer
    model.add(keras.layers.Flatten())
    model.add(keras.layers.Dense(Nlabels))
    model.add(keras.layers.Activation("softmax"))

    model.compile(loss = "sparse_categorical_crossentropy",
                  optimizer = "adam",
                  metrics = ["accuracy"])

    training = model.fit(x_train, y_train,
                                           batch_size = 32,
                                           epochs = Nepochs,
                                           validation_data = (x_test, y_test),
                                           shuffle = True)
    histories[Nepochs, deep] = training.history
```
