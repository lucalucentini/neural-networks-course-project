```python
import numpy as np
import tensorflow as tf
from tensorflow import keras
import matplotlib.pyplot as plt
mnist = keras.datasets.mnist
(x_train, y_train), (x_test, y_test) = mnist.load_data()

# Normalizzazione
x_train = x_train/255
x_test = x_test/255
```

```python
model = keras.models.Sequential()

model.add(keras.layers.Flatten()) # input layer: appiattisce l'immagine in una colonna
model.add(keras.layers.Dense(128, activation = tf.nn.relu))
model.add(keras.layers.Dense(128, activation = tf.nn.relu))   #2 hidden layers
model.add(keras.layers.Dense(10, activation = tf.nn.softmax))   # output

model.compile(optimizer = 'adam',
              loss = 'sparse_categorical_crossentropy',
              metrics = ['accuracy'])
```

```python
early_stopping = keras.callbacks.EarlyStopping(
    monitor = 'loss',             # metrica da controllare
    patience = 2,                 # quante epoche tollerare senza miglioramento
    min_delta = 1e-4,
    restore_best_weights = True   # ripristina i pesi migliori
)

Nepochs = 10
training1 = model.fit(x_train, y_train,
                      epochs = Nepochs,
                      batch_size = 32,
                      callbacks = [early_stopping],
                      shuffle = True,
                      validation_data = (x_test, y_test))


val_loss, val_acc = model.evaluate(x_test, y_test)
print(training1.history['loss'][-1], val_loss, training1.history['accuracy'][-1], val_acc)
```

```python
plt.plot(training1.history['loss'], label = 'loss')
plt.plot(training1.history['val_loss'], label = 'val_loss')
plt.plot(training1.history['accuracy'], label = 'accuracy')
plt.plot(training1.history['val_accuracy'], label = 'val_accuracy')

plt.legend()
plt.show()
```

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay

cm = confusion_matrix(y_test, np.argmax(model.predict(x_test), axis = 1))
disp = ConfusionMatrixDisplay(confusion_matrix = cm)
disp.plot(cmap = "Blues")
```
