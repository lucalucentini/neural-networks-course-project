```python
import numpy as np
def training_set(tipologia):
    P = 10
    if tipologia == 'due_parabole':

        # Numero di punti
        N = (P + 1) ** 2

        # Inizializzazione dei punti e delle classi
        x1 = np.zeros(N)
        x2 = np.zeros(N)
        y = np.zeros((3, N))

        # Definizione delle due parabole
        p1 = lambda x: 2 * x * (1 - x)
        p2 = lambda x: 2 * x * (1 - x) + 0.5

        # Distribuzione dei punti sulla griglia e assegnazione delle classi
        k = 0
        for i in range(P + 1):
            for j in range(P + 1):
                x1[k] = i / P
                x2[k] = j / P

                if x2[k] > p2(x1[k]):
                    y[:, k] = [1, 0, 0]
                elif x2[k] < p1(x1[k]):
                    y[:, k] = [0, 0, 1]
                else:
                    y[:, k] = [0, 1, 0]
                k = k + 1

    elif tipologia == 'due_parabole_random':

        # Numero di punti
        N = (P + 1) ** 2

        # Inizializzazione dei punti e delle classi
        x1 = np.zeros(N)
        x2 = np.zeros(N)
        y = np.zeros((3, N))

        # Definizione delle due parabole
        p1 = lambda x: 2 * x * (1 - x)


        p2 = lambda x: 2 * x * (1 - x) + 0.5

        # Distribuzione casuale dei punti e assegnazione delle classi
        k = 0
        for i in range(P + 1):
            for j in range(P + 1):
                x1[k] = np.random.rand()
                x2[k] = np.random.rand()

                if x2[k] > p2(x1[k]):
                    y[:, k] = [1, 0, 0]
                elif x2[k] < p1(x1[k]):
                    y[:, k] = [0, 0, 1]
                else:
                    y[:, k] = [0, 1, 0]
                k = k + 1

    elif tipologia == 'cerchio':
        
        # Numero di punti
        N = (P + 1) ** 2
        
        # Inizializzazione dei punti e delle classi (2 classi)
        x1 = np.zeros(N)
        x2 = np.zeros(N)
        y  = np.zeros((2, N))

        # Definizione del cerchio: centro e raggio
        xc = 0.5
        yc = 0.5
        r  = 0.25

        # Distribuzione dei punti sulla griglia e assegnazione delle classi
        k = 0
        for i in range(P + 1):
            for j in range(P + 1):
                x1[k] = i / P
                x2[k] = j / P

                if (x1[k] - xc)**2 + (x2[k] - yc)**2 < r**2:
                    # dentro il cerchio
                    y[:, k] = [1, 0]
                else:
                    # fuori dal cerchio
                    y[:, k] = [0, 1]
                k = k + 1

    elif tipologia == 'cerchio_random':
        
        # Numero di punti
        N = (P + 1) ** 2
        
        # Inizializzazione dei punti e delle classi (2 classi)
        x1 = np.zeros(N)
        x2 = np.zeros(N)
        y  = np.zeros((2, N))

        # Definizione del cerchio: centro e raggio
        xc = 0.5
        yc = 0.5
        r  = 0.25

        # Distribuzione casuale dei punti e assegnazione delle classi
        k = 0
        for i in range(P + 1):
            for j in range(P + 1):
                x1[k] = np.random.rand()
                x2[k] = np.random.rand()

                if (x1[k] - xc)**2 + (x2[k] - yc)**2 < r**2:
                    # dentro il cerchio
                    y[:, k] = [1, 0]
                else:
                    # fuori dal cerchio
                    y[:, k] = [0, 1]
                k = k + 1
    elif tipologia == 'due_rette':
        N = (P + 1) ** 2
        x1 = np.zeros(N); x2 = np.zeros(N); y = np.zeros((3, N))
        p1 = lambda x: 1/3
        p2 = lambda x: 2/3 + 0.2 * x
        k = 0
        for i in range(P + 1):
            for j in range(P + 1):
                x1[k] = i / P; x2[k] = j / P
                if x2[k] > p2(x1[k]):      
                    y[:, k] = [1, 0, 0]
                elif x2[k] < p1(x1[k]):   
                    y[:, k] = [0, 0, 1]
                else:                      
                    y[:, k] = [0, 1, 0]
                k = k + 1

    elif tipologia == 'due_rette_random':
        N = (P + 1) ** 2
        x1 = np.zeros(N); x2 = np.zeros(N); y = np.zeros((3, N))
        p1 = lambda x: 1/3
        p2 = lambda x: 2/3 + 0.2 * x
        k = 0
        for i in range(P + 1):
            for j in range(P + 1):
                x1[k] = np.random.rand(); x2[k] = np.random.rand()
                if x2[k] > p2(x1[k]):      
                    y[:, k] = [1, 0, 0]
                elif x2[k] < p1(x1[k]):   
                    y[:, k] = [0, 0, 1]
                else:                      
                    y[:, k] = [0, 1, 0]
                k = k + 1
    else:
        raise ValueError("tipologia non riconosciuta: usa 'due_parabole', 'due_parabole_random', ... ")

    return x1, x2, y
```

```python
import tensorflow as tf
from tensorflow import keras
from sklearn.model_selection import train_test_split
```

```python
def keras_nn(nlv, act_fn, optimizer_name, eta, loss):

    model = keras.models.Sequential()

    # Input layer
    model.add(keras.layers.Input(shape=(nlv[0],)))



    init = keras.initializers.GlorotUniform()
    #alternativa: init = keras.initializers.RandomNormal(mean = 0.0, stddev = 1.0)

    for neuroni in nlv[1:-1]:
        model.add(keras.layers.Dense(neuroni,
                                     activation = act_fn,
                                     kernel_initializer = init,
                                     bias_initializer = init))          # Hidden layers

    model.add(keras.layers.Dense(nlv[-1],
                                 activation = 'softmax',
                                 kernel_initializer = init,
                                 bias_initializer = init))          # Output layer

    # Definiamo gli optimizers
    if optimizer_name == "SGD":
        optimizer = keras.optimizers.SGD(learning_rate=eta)

    elif optimizer_name == "Adam":
        optimizer = keras.optimizers.Adam(learning_rate=eta)

    else:
        raise ValueError("Aggiungere {} a keras_nn".format(optimizer_name))

    # Configurazione pre-training del modello
    model.compile(optimizer = optimizer,
                  loss = loss,
                  metrics = ['CategoricalAccuracy'])

    return model
```

```python
# Preparazione del dataset
tf.random.set_seed(0)

x1, x2, y = training_set('due_parabole')

x = np.column_stack([x1, x2])
y = y.T

x_tr, x_val, y_tr, y_val = train_test_split(
    x, y, test_size = 0.2, random_state = 0, shuffle = True, stratify = np.argmax(y, axis=1)
)

# Definizione dell'architettura
layers = np.array([len(np.array((x1[0], x2[0]))), 5, 9, 8, len(y[0,:])])

#Definiamo la loss
def my_loss(y_true, y_pred):


    return 0.5 * tf.reduce_sum((y_pred - y_true) ** 2, axis = -1)              # tf.reduce_sum() <--> np.sum

# Definizione del modello
model = keras_nn(nlv = layers,
                 act_fn = 'sigmoid',
                 optimizer_name = 'SGD',
                 eta = 0.05,
                 loss = my_loss)

Nepochs = 8000

# Eventuale early stopping
early_stopping = keras.callbacks.EarlyStopping(
    monitor = 'val_CategoricalAccuracy',       # metrica da controllare
    mode = 'max',
    patience = 700,              # quante epoche tollerare senza miglioramento
    restore_best_weights = True   # ripristina i pesi migliori
)
```

```python
# Addestramento
training = model.fit(x_tr, y_tr,
                     validation_data = (x_val, y_val),
                     batch_size = 32,
                     epochs = Nepochs,
                     shuffle = True,
                     callbacks = [early_stopping])
```

```python
import matplotlib.pyplot as plt
plt.plot(training.history['loss'], label = 'Training loss')
plt.plot(training.history['val_loss'], label = 'Validation loss')
plt.plot(training.history['CategoricalAccuracy'], label = 'Training accuracy')
plt.plot(training.history['val_CategoricalAccuracy'], label = 'Validation accuracy')

plt.legend()
plt.show()
```
