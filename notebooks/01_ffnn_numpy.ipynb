```python
import numpy as np
def training_set(tipologia, P=10):
    if tipologia == 'due_parabole':

        # Numero di punti
        N = (P + 1) ** 2

        # Inizializzazione dei punti e delle classi
        x1 = np.zeros(N)
        x2 = np.zeros(N)
        y = np.zeros((3, N))

        # Definizione delle due parabole
        p1 = lambda x: 2 * x * (1 - x)
        p2 = lambda x: 2 * x * (1 - x) + 0.5

        # Distribuzione dei punti sulla griglia e assegnazione delle classi
        k = 0
        for i in range(P + 1):
            for j in range(P + 1):
                x1[k] = i / P
                x2[k] = j / P

                if x2[k] > p2(x1[k]):
                    y[:, k] = [1, 0, 0]
                elif x2[k] < p1(x1[k]):
                    y[:, k] = [0, 0, 1]
                else:
                    y[:, k] = [0, 1, 0]
                k = k + 1

    elif tipologia == 'due_parabole_random':

        # Numero di punti
        N = (P + 1) ** 2

        # Inizializzazione dei punti e delle classi
        x1 = np.zeros(N)
        x2 = np.zeros(N)
        y = np.zeros((3, N))

        # Definizione delle due parabole
        p1 = lambda x: 2 * x * (1 - x)


        p2 = lambda x: 2 * x * (1 - x) + 0.5

        # Distribuzione casuale dei punti e assegnazione delle classi
        k = 0
        for i in range(P + 1):
            for j in range(P + 1):
                x1[k] = np.random.rand()
                x2[k] = np.random.rand()

                if x2[k] > p2(x1[k]):
                    y[:, k] = [1, 0, 0]
                elif x2[k] < p1(x1[k]):
                    y[:, k] = [0, 0, 1]
                else:
                    y[:, k] = [0, 1, 0]
                k = k + 1

    elif tipologia == 'cerchio':
        
        # Numero di punti
        N = (P + 1) ** 2
        
        # Inizializzazione dei punti e delle classi (2 classi)
        x1 = np.zeros(N)
        x2 = np.zeros(N)
        y  = np.zeros((2, N))

        # Definizione del cerchio: centro e raggio
        xc = 0.5
        yc = 0.5
        r  = 0.25

        # Distribuzione dei punti sulla griglia e assegnazione delle classi
        k = 0
        for i in range(P + 1):
            for j in range(P + 1):
                x1[k] = i / P
                x2[k] = j / P

                if (x1[k] - xc)**2 + (x2[k] - yc)**2 < r**2:
                    # dentro il cerchio
                    y[:, k] = [1, 0]
                else:
                    # fuori dal cerchio
                    y[:, k] = [0, 1]
                k = k + 1

    elif tipologia == 'cerchio_random':
        
        # Numero di punti
        N = (P + 1) ** 2
        
        # Inizializzazione dei punti e delle classi (2 classi)
        x1 = np.zeros(N)
        x2 = np.zeros(N)
        y  = np.zeros((2, N))

        # Definizione del cerchio: centro e raggio
        xc = 0.5
        yc = 0.5
        r  = 0.25

        # Distribuzione casuale dei punti e assegnazione delle classi
        k = 0
        for i in range(P + 1):
            for j in range(P + 1):
                x1[k] = np.random.rand()
                x2[k] = np.random.rand()

                if (x1[k] - xc)**2 + (x2[k] - yc)**2 < r**2:
                    # dentro il cerchio
                    y[:, k] = [1, 0]
                else:
                    # fuori dal cerchio
                    y[:, k] = [0, 1]
                k = k + 1
    elif tipologia == 'due_rette':
        N = (P + 1) ** 2
        x1 = np.zeros(N); x2 = np.zeros(N); y = np.zeros((3, N))
        p1 = lambda x: 1/3
        p2 = lambda x: 2/3 + 0.2 * x
        k = 0
        for i in range(P + 1):
            for j in range(P + 1):
                x1[k] = i / P; x2[k] = j / P
                if x2[k] > p2(x1[k]):      
                    y[:, k] = [1, 0, 0]
                elif x2[k] < p1(x1[k]):   
                    y[:, k] = [0, 0, 1]
                else:                      
                    y[:, k] = [0, 1, 0]
                k = k + 1

    elif tipologia == 'due_rette_random':
        N = (P + 1) ** 2
        x1 = np.zeros(N); x2 = np.zeros(N); y = np.zeros((3, N))
        p1 = lambda x: 1/3
        p2 = lambda x: 2/3 + 0.2 * x
        k = 0
        for i in range(P + 1):
            for j in range(P + 1):
                x1[k] = np.random.rand(); x2[k] = np.random.rand()
                if x2[k] > p2(x1[k]):      
                    y[:, k] = [1, 0, 0]
                elif x2[k] < p1(x1[k]):   
                    y[:, k] = [0, 0, 1]
                else:                      
                    y[:, k] = [0, 1, 0]
                k = k + 1
    else:
        raise ValueError("tipologia non riconosciuta: usa 'due_parabole', 'due_parabole_random', ... ")

    return x1, x2, y
```

```python
def my_activate(x, W, b, activation):
    z = W @ x + b
    if activation == 'linear':
        y = z                      # attivazione
        dy = np.ones(len(z))       # derivata attivazione
    elif activation == 'sigmoid':
        y = 1 / (1 + np.exp(-z))
        dy = y * (1-y)
    elif activation == 'relu':
        y = np.maximum(0, z)
        dy = (z > 0).astype(float)
    else:
        raise ValueError('Attivazione non supportata')
    return [y,dy]
```

```python
def F(x, Wb, nlv, act_fn):
    a = np.asarray(x).reshape(-1, 1)
    for l in range(1, len(nlv)):
        a, derivate = my_activate(a, Wb['weights'][l], Wb['biases'][l], act_fn)
    return a.reshape(-1)
```

```python
# Definizione e addestramento della rete
def nn_train(x1, x2, y, act_fn, nlv, M, Nepochs, eta, TOL = 1e-4):

    N = len(x1)                # N vettori di input della forma x = (x1, x2) \in \R^2
    L = len(nlv)               # L è il numero di layers, compreso l'input layer

    Wb = {'weights':[0], 'biases':[0]}

    # Inizializzazione dei parametri
    for l in range(1,L):
       Wb['weights'].append(np.random.normal(0,1,(nlv[l],nlv[l-1])))
       Wb['biases'].append(np.random.normal(0,1,(nlv[l],1)))

    checkpoint = Nepochs // 100            # Ogni quante epoche stampare il costo

    for n in range(Nepochs):

        rkv = np.random.permutation(N)           # Permutazione degli indici 0, ..., N-1

        for b in range(0, N, M):          # 1 completamento del ciclo = 1 epoca
            k = rkv[b:b+M]                # Mini-batch corrente (senza reinserimento)
            if len(k) < M:                # Salta l'ultimo mini-batch se incompleto
                continue

            grad_C = {'weights':[[None for i in range(M)] for j in range(L)],
                    'biases':[[None for i in range(M)] for j in range(L)]}

            for m in range(M):
                x = np.array([x1[k[m]], x2[k[m]]]).reshape(-1, 1)               #data point

                # Forward pass
                a = [None for l in range(L)]
                a[0] = np.asarray(x).reshape(-1, 1)
                D = [None for l in range(L)]
                D[0] = np.identity(len(x))

                for l in range(1,L):
                    act_y, act_dy = my_activate(x = a[l-1], W = Wb['weights'][l],
                                                b = Wb['biases'][l], activation = act_fn)

                    a[l]    = np.asarray(act_y).reshape(-1, 1) # attivazioni come colonna


                    act_dy = np.asarray(act_dy).reshape(-1)      # 1-D prima di fare diag
                    D[l]   = np.diag(act_dy)

                # Backward pass
                delta = [None for l in range(L)]
                delta[L-1] = D[L-1] @ (a[L-1] - y[:, k[m]].reshape(-1, 1))

                for l in range(L-2, 0, -1):
                    delta[l] = D[l] @ (Wb['weights'][l+1].T @ delta[l+1])

                # Salviamo i gradienti
                for l in range(L-1, 0, -1):
                    grad_C['weights'][l][m] = delta[l] @ a[l-1].T
                    grad_C['biases'][l][m] = delta[l]

            # Gradient step
            sum_grad_C = {'weights':[None for i in range(L)],
                        'biases':[None for i in range(L)]}
            mean_grad_C = {'weights':[None for i in range(L)],
                        'biases':[None for i in range(L)]}

            for l in range(1,L):
                sum_grad_C['weights'][l] = np.zeros((nlv[l], nlv[l-1]))
                sum_grad_C['biases'][l] = np.zeros((nlv[l], 1))

                for m in range(M):
                    sum_grad_C['weights'][l] = (
                        sum_grad_C['weights'][l]
                        + grad_C['weights'][l][m]
                    )
                    sum_grad_C['biases'][l] = (
                        sum_grad_C['biases'][l]
                        + grad_C['biases'][l][m]
                    )

                mean_grad_C['weights'][l] = sum_grad_C['weights'][l]/M
                mean_grad_C['biases'][l] = sum_grad_C['biases'][l]/M

                #Optimization step: aggiornamento dei parametri
            for l in range(1,L):
                Wb['weights'][l] = Wb['weights'][l] - eta * mean_grad_C['weights'][l]
                Wb['biases'][l] = Wb['biases'][l] - eta * mean_grad_C['biases'][l]

    #cost_sum = 0.0

    #for i in range(N):
        #a = F([x1[i], x2[i]], Wb, nlv, act_fn)
        #diff = a - y[:, i].reshape(-1, 1)
        #cost_sum = cost_sum + 0.5 * np.sum(diff**2)
    #cost_avg = cost_sum / N

    #if n % checkpoint == 0:
        #print("Cost = {}, progresso = {}% ".format(cost_avg, ((n+1) / Nepochs) * 100))

    # EARLY STOPPING
    #if cost_avg <= TOL:
            #break

    return Wb
```

```python
def previsione(x1, x2, y, Wb, nlv, act_fn):
    N = len(x1)

    y_true = y.copy()                           # (y.shape[0], N)
    y_score = np.zeros((y.shape[0], N))         # (y.shape[0], N)
    y_pred = np.zeros((y.shape[0], N))          # (y.shape[0], N)
    for i in range(N):
        output_rete = F([x1[i], x2[i]], Wb, nlv, act_fn).reshape(-1)         # (y.shape[0],)
        y_score[:, i] = output_rete
        y_pred[output_rete.argmax(), i] = 1
    return y_score, y_pred, y_true
```

```python
def regioni(x1, x2, y, Wb, nlv, act_fn, F, ngrid = 100):

    # Creazione delle liste vuote da riempire con i punti classificati
    xxA, xxB, xxC = [], [], []

    # Creazione griglia di punti e assegnazione alla classe predetta da F
    for u in np.linspace(0, 1, ngrid):
        for v in np.linspace(0, 1, ngrid):
            a = np.array([u, v]).reshape(-1, 1)             # input 2x1
            out = F(a, Wb, nlv, act_fn).reshape(-1, 1)      # output 3x1
            j = np.argmax(out)                              # 0,1,2, classe predetta
            (xxA if j == 0 else xxB if j == 1 else xxC).append([u, v])

    # Creazione grafico delle regioni di classificazione
    plt.figure(figsize = (6,6))
    if len(xxA) > 0: plt.scatter(np.array(xxA)[:, 0], np.array(xxA)[:, 1], s=5, c='r', marker='.')
    if len(xxB) > 0: plt.scatter(np.array(xxB)[:, 0], np.array(xxB)[:, 1], s=5, c='g', marker='.')
    if len(xxC) > 0 : plt.scatter(np.array(xxC)[:, 0], np.array(xxC)[:, 1], s=5, c='b', marker='.')

    # Aggiungiamo il dataset al grafico
    for i in range(len(x1)):
        j = np.argmax(y[:, i]) # classe vera
        marker = {0: 'ok', 1: 'xk', 2: 'dk'}[j]
        plt.plot(x1[i], x2[i], marker, linewidth = 2)

    plt.show()
```

```python
def grid_label(tipologia, u, v):
    if tipologia == 'due_parabole' or tipologia == 'due_parabole_random':
        p1 = lambda x: 2 * x * (1 - x)
        p2 = lambda x: 2 * x * (1 - x) + 0.5
        if v > p2(u):
            return np.array([1., 0., 0.])
        elif v < p1(u):
            return np.array([0., 0., 1.])
        else:
            return np.array([0., 1., 0.])

    if tipologia == 'due_rette' or tipologia == 'due_rette_random':
        p1 = lambda x: 1/3
        p2 = lambda x: 2/3 + 0.2 * x
        if v > p2(u):
            return np.array([1., 0., 0.])
        elif v < p1(u):
            return np.array([0., 0., 1.])
        else:
            return np.array([0., 1., 0.])

    if tipologia == 'cerchio' or tipologia == 'cerchio_random':
        xc, yc, r = 0.5, 0.5, 0.25
        if (u - xc) ** 2 + (v - yc) ** 2 < r ** 2:
            return np.array([1., 0.])
        else:
            return np.array([0., 1.])

    raise ValueError("Tipologia non supportata")
```

```python
# Accuracy e loss sulla griglia

def grid_accuracy(tipologia, Wb, nlv, act_fn, F, ngrid=100):
    us = np.linspace(0, 1, ngrid)
    vs = np.linspace(0, 1, ngrid)

    corrette = 0
    for u in us:
        for v in vs:
            y_pred = np.argmax(F((u, v), Wb, nlv, act_fn))
            y_true = np.argmax(grid_label(tipologia, u, v))
            corrette = corrette + (y_pred == y_true)
    return corrette / (ngrid * ngrid)

# Loss sulla griglia
def grid_loss(tipologia, Wb, nlv, act_fn, F, ngrid=100):
    us = np.linspace(0, 1, ngrid)
    vs = np.linspace(0, 1, ngrid)

    loss = 0
    for u in us:
        for v in vs:
             differenza = F((u, v), Wb, nlv, act_fn).reshape(-1) - grid_label(tipologia, u, v)
             loss = loss + 0.5 * np.sum(differenza ** 2)
    return loss / (ngrid * ngrid)
```

```python
configurazioni = [
    ('due_parabole_random', 10, [4, 4]),
    ('due_parabole', 10, [4, 4]),
    ('due_parabole', 20, [5, 9, 8]),
    ('due_parabole_random', 20, [5, 9, 8]),
    ('cerchio', 20, [5, 9, 8]),
    ('cerchio_random', 20, [5, 9, 8]),
    ('due_rette', 20, [5, 9, 8]),
    ('due_rette_random', 20, [5, 9, 8]),
]

for tipologia, P, hidden in configurazioni:
    print(tipologia, (P + 1) ** 2, hidden)
    # Addestriamo la rete su due_parabole_random
    np.random.seed(0)
    x1, x2, y = training_set(tipologia, P)
    layers = np.array([len(np.array((x1[0], x2[0]))), *hidden, len(y[:,0])])

    parametri = nn_train(x1, x2, y, act_fn = 'sigmoid',
                  nlv = layers,
                  M = 5, Nepochs = 10 ** 4, eta = 0.05)

    y_score, y_pred, y_true = previsione(x1, x2, y, parametri, nlv = layers, act_fn = 'sigmoid')
    regioni(x1, x2, y, parametri, nlv = layers, act_fn = 'sigmoid', F = F, ngrid = 100)

    # Accuracy sul training set
    accuracy = np.mean(y_pred.argmax(axis = 0) == y_true.argmax(axis = 0))
    print(accuracy)

    costo_finale = (1/len(x1)) * 0.5 * np.sum((([F([x1[i], x2[i]], parametri, layers, 'sigmoid')
                                                 for i in range(len(x1))]) - y_true.T) ** 2)
    print(costo_finale)

    # Accuracy sulla griglia

    print(grid_accuracy(tipologia, parametri, layers, 'sigmoid', F = F, ngrid = 100))

    print(grid_loss(tipologia, parametri, layers, 'sigmoid', F = F, ngrid = 100))
```
