```python
import numpy as np
import tensorflow as tf
from tensorflow import keras
import matplotlib.pyplot as plt

tf.random.set_seed(0)

# Importiamo i dati
mnist = keras.datasets.mnist
(x_train, y_train), (x_test, y_test) = mnist.load_data()

# Normalizzazione
x_train = x_train/255
x_test = x_test/255

Nlabels = np.max(y_train) + 1           # In questo caso le classi sono {0, ..., Nlabels - 1}

Nepochs = 6

model = keras.models.Sequential()

# Convolutional layer
model.add(keras.layers.Input(shape = (28, 28, 1)))
model.add(keras.layers.Conv2D(64, (3,3)))
model.add(keras.layers.Activation("relu"))

# Pooling layer
model.add(keras.layers.MaxPooling2D(pool_size = (2,2)))



# Fully connected layer
model.add(keras.layers.Flatten())
model.add(keras.layers.Dense(Nlabels))
model.add(keras.layers.Activation("softmax"))

model.compile(loss = "sparse_categorical_crossentropy",
              optimizer = "adam",
              metrics = ["accuracy"])

training_convnet_mnist = model.fit(x_train, y_train,
                             batch_size = 32,
                             epochs = Nepochs,
                             validation_data = (x_test, y_test),
                             shuffle = True)
```

```python
plt.plot(training_convnet_mnist.history['loss'], label = 'loss')
plt.plot(training_convnet_mnist.history['val_loss'], label = 'val_loss')
plt.plot(training_convnet_mnist.history['accuracy'], label = 'accuracy')
plt.plot(training_convnet_mnist.history['val_accuracy'], label = 'val_accuracy')

plt.legend()
plt.show()
```
